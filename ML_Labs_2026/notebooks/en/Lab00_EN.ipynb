```python
distances = [1, 2, 3]
print(distances[0])
print(len(distances))
```

A function groups a calculation. Its indented body uses four spaces. return sends the result back to the caller.

```python
def predict_minutes(x):
    return 10 * x + 1

print(predict_minutes(distance))
```

```python
for x in distances:
    print(x, predict_minutes(x))
```
