## A2 · Understand the idea

The affine model predicts y_hat = w*x + b. Its slope w has unit min/km and its intercept b has unit min. The residual is prediction minus observation: e_i = y_hat_i - y_i. Absolute error is |e_i|. MAE = (|e_1| + ... + |e_n|)/n, expressed in minutes.

LinearRegression fits w and b by minimizing the sum of squared training residuals, with an intercept. Evaluation here uses validation MAE. The baseline predicts the training mean for every validation journey. Fitting and evaluation therefore have distinct roles. A change to a manual line changes its predictions; training estimates parameters from the training data.

## A3 · Worked example

```python
observed = np.array([12., 19., 32.])
predicted = np.array([10., 20., 30.])
absolute_errors = np.abs(predicted - observed)
print(absolute_errors, absolute_errors.mean())
from sklearn.linear_model import LinearRegression
sets = split_data("walks.csv")
X_train, y_train = sets["train"]
model = LinearRegression().fit(X_train, y_train)
print("w, b:", model.coef_[0], model.intercept_)
print("2 km:", model.predict([[2.]]))
```

Absolute errors are [2, 1, 2], hence MAE = 5/3 ≈ 1.667 min. In the supplied reference run, the fitted slope is about 11.281 and validation MAE about 2.569 min. Read the exact numbers and software versions in the result page.
