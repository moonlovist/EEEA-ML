## A2 · Understand the idea

The fixed fictional classification data contain two numerical coordinates and a label, 0 or 1. There are 96 training, 48 validation and 48 reserved test observations. kNN assigns the majority class among k nearby training points. The pipeline estimates feature means and scales on training data before measuring distances. A decision tree successively compares features with thresholds. Its maximum depth limits the number of comparisons along a path.

Accuracy = correct predictions / evaluated observations. A confusion matrix uses true labels as rows and predicted labels as columns in this course. Diagonal entries count correct predictions; off-diagonal entries describe mistakes. The plot background shows predicted classes, while point colours indicate observed classes.

## A3 · Worked example

```python
from sklearn.metrics import accuracy_score, confusion_matrix
truth = [0, 0, 0, 1, 1, 1]
prediction = [0, 0, 1, 1, 1, 1]
print("accuracy:", accuracy_score(truth, prediction))
print(confusion_matrix(truth, prediction, labels=[0, 1]))
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
sets = split_data()
knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
knn.fit(*sets["train"])
print("validation accuracy:", knn.score(*sets["validation"]))
```

The six-example accuracy is 5/6 ≈ 0.8333. Its matrix is [[2,1],[0,3]]: one true class-0 example was predicted as class 1. The supplied k=5 run has 41/48 correct validation predictions.
