## A2 · Understand the idea

Raw squared distance is the sum of coordinate differences squared. Here the contributions are 1² and 100², giving 10001. Standardisation transforms x to z=(x−mean)/sd. StandardScaler uses population standard deviation: sd = sqrt(sum((x_i−mean)²)/n). On [1,3], mean=2, sd=1, so 3 becomes 1.

K-means alternates assignment to a nearest centre and replacement of each centre by its assigned mean. Its objective, inertia, is the sum of squared within-cluster distances. The chosen scaling defines the geometry. Compare inertia values within the same feature space. Silhouette also depends on that geometry. Cluster IDs are arbitrary names. Here clustering explores all 90 fictional customers; assessing future-customer assignment would require a separate held-out protocol.

## A3 · Worked example

```python
values = np.array([1., 3.])
mean = values.mean()
sd = values.std(ddof=0)
print("mean, sd:", mean, sd)
print("z:", (values - mean) / sd)
print("raw squared distance:", 1**2 + 100**2)
from sklearn.preprocessing import StandardScaler
rows = read_rows("customers.csv")
X = np.array([[float(r["visits_month"]), float(r["spend_eur_month"])] for r in rows])
scaler = StandardScaler().fit(X)
print("standardised means:", scaler.transform(X).mean(axis=0))
```

The two standardised values are −1 and 1. Standardised feature means are approximately zero; tiny floating-point residuals are expected. Plot axes and reported centres remain in visits/month and EUR/month for interpretation.
