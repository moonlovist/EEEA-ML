## A2 · Understand the idea

Model complexity controls the set of patterns a model can express. Increasing tree depth can improve its fit to individual training observations. Generalisation concerns performance on further observations from the relevant population. A gap between high training performance and weaker validation performance is evidence to examine for overfitting.

The selection rule here is fixed in advance: highest validation accuracy, with the shallower candidate selected in a tie. Standardisation parameters and other fitted preprocessing belong to the training stage. A feature must be available at the intended prediction time: the final observed duration would reveal the answer when predicting whether a journey exceeds 25 minutes. That is target leakage.

## A3 · Worked example

```python
from sklearn.tree import DecisionTreeClassifier
sets = split_data()
for depth in [1, 2, 3, 5, 10]:
    tree = DecisionTreeClassifier(max_depth=depth, random_state=7)
    tree.fit(*sets["train"])
    print(depth, tree.score(*sets["train"]), tree.score(*sets["validation"]))
```

The supplied curve reaches training accuracy 1.0 at large depth. The first maximum validation accuracy occurs at depth 2; depth 3 ties it. The fixed tie rule selects depth 2. Read the actual table if your environment produces a different measured result.
