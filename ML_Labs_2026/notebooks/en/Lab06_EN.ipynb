## A2 · Understand the idea

Our MLP has two inputs, eight tanh hidden units and one sigmoid output for class-1 probability. Its 33 parameters are 2×8 input weights, 8 hidden biases, 8 output weights and 1 output bias. Forward computation is a=XW1+b1, h=tanh(a), z=hW2+b2, p=sigmoid(z).

For labels y in {0,1}, mean binary cross-entropy is L=−mean[y log(p)+(1−y)log(1−p)]. The code evaluates the equivalent stable expression mean[log(1+exp(z))−yz]. The chain rule gives dL/dz=(p−y)/n. Then dW2=hᵀ(dL/dz), db2=sum(dL/dz), da=((dL/dz)W2ᵀ)⊙(1−h²), dW1=Xᵀda and db1=sum_rows(da). Each parameter updates as parameter−rate×gradient, using gradients computed at the same old parameter values. Each update uses all 96 training rows. Scaling is fitted on those rows.
