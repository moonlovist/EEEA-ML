## A2 · Comprendre l’idée

Le MLP possède deux entrées, huit unités cachées tanh et une sortie sigmoïde donnant la probabilité de classe 1. Ses 33 paramètres comprennent 2×8 poids d’entrée, 8 biais cachés, 8 poids de sortie et 1 biais final. Le calcul direct est a=XW1+b1, h=tanh(a), z=hW2+b2, p=sigmoid(z).

Pour y dans {0,1}, l’entropie croisée binaire moyenne vaut L=−mean[y log(p)+(1−y)log(1−p)]. Le code évalue la forme stable équivalente mean[log(1+exp(z))−yz]. La règle de chaîne donne dL/dz=(p−y)/n. Puis dW2=hᵀ(dL/dz), db2=sum(dL/dz), da=((dL/dz)W2ᵀ)⊙(1−h²), dW1=Xᵀda et db1=sum_rows(da). Chaque paramètre devient paramètre−pas×gradient, avec tous les gradients calculés aux mêmes anciennes valeurs. Chaque mise à jour utilise les 96 lignes d’entraînement ; la standardisation est ajustée sur ces lignes.
