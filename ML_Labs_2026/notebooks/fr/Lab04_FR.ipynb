## A3 · Exemple guidé

```python
from sklearn.tree import DecisionTreeClassifier
sets = split_data()
for depth in [1, 2, 3, 5, 10]:
    tree = DecisionTreeClassifier(max_depth=depth, random_state=7)
    tree.fit(*sets["train"])
    print(depth, tree.score(*sets["train"]), tree.score(*sets["validation"]))
```

La courbe fournie atteint 1,0 sur l’entraînement à grande profondeur. Le premier maximum de validation apparaît à la profondeur 2 ; la profondeur 3 est à égalité. La règle fixe retient 2. Lire le tableau effectif si votre environnement produit un résultat différent.
