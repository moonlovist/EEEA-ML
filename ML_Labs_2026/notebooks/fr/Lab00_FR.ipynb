Une cellule de code contient des instructions. Maj+Entrée exécute la cellule sélectionnée. Le numéro à côté indique l’ordre d’exécution. Lire la sortie prédite avant de lancer.

Le symbole = affecte une valeur à une variable. print l’affiche. La multiplication utilise *.

```python
distance = 2
prediction = 10 * distance + 1
print(prediction)
```

Une liste contient plusieurs valeurs. L’indice 0 sélectionne le premier élément. len compte les éléments.

```python
distances = [1, 2, 3]
print(distances[0])
print(len(distances))
```

Une fonction regroupe un calcul. Son corps indenté utilise quatre espaces. return renvoie le résultat au code appelant.

```python
def predict_minutes(x):
    return 10 * x + 1

print(predict_minutes(distance))
```

Une boucle répète le calcul pour chaque élément. Expliquer une itération à voix haute.

```python
for x in distances:
    print(x, predict_minutes(x))
```

Un tableau possède une forme. Ici (3,1) signifie trois observations et une variable. X[:,0] sélectionne toutes les lignes de la première colonne.

```python
import numpy as np
X = np.array([[1.], [2.], [3.]])
print(X.shape)
print(X[:, 0])
```
