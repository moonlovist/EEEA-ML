## A2 · Comprendre l’idée

Les données fictives fixées contiennent deux coordonnées numériques et une étiquette 0 ou 1. Elles comprennent 96 observations d’entraînement, 48 de validation et 48 réservées au test. kNN attribue la classe majoritaire parmi k points d’entraînement proches. Le pipeline estime moyennes et échelles sur l’entraînement avant de mesurer les distances. Un arbre compare successivement des variables à des seuils. Sa profondeur maximale borne le nombre de comparaisons sur un chemin.

Exactitude = prédictions correctes / observations évaluées. Dans ce cours, les lignes de la matrice de confusion représentent les vraies classes et les colonnes les classes prédites. La diagonale compte les réussites ; les autres cases décrivent les erreurs. Le fond du graphique représente les classes prédites ; les couleurs des points indiquent les classes observées.

## A3 · Exemple guidé

```python
from sklearn.metrics import accuracy_score, confusion_matrix
truth = [0, 0, 0, 1, 1, 1]
prediction = [0, 0, 1, 1, 1, 1]
print("accuracy:", accuracy_score(truth, prediction))
print(confusion_matrix(truth, prediction, labels=[0, 1]))
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
sets = split_data()
knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
knn.fit(*sets["train"])
print("validation accuracy:", knn.score(*sets["validation"]))
```

L’exactitude sur six exemples vaut 5/6 ≈ 0,8333. La matrice est [[2,1],[0,3]] : un exemple de classe 0 est prédit en classe 1. L’exécution fournie avec k=5 réussit 41 prédictions de validation sur 48.
