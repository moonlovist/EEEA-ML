## A2 · Comprendre l’idée

La distance brute au carré est la somme des différences de coordonnées au carré. Les contributions valent ici 1² et 100², soit 10001. La standardisation transforme x en z=(x−moyenne)/écart-type. StandardScaler utilise l’écart-type de population : sd = sqrt(sum((x_i−moyenne)²)/n). Pour [1,3], la moyenne vaut 2, sd vaut 1 et 3 devient 1.

K-means alterne affectation au centre le plus proche et remplacement de chaque centre par la moyenne des points affectés. Son objectif, l’inertie, est la somme des distances internes au carré. L’échelle choisie définit la géométrie. Comparer les inerties dans un même espace de variables. La silhouette dépend aussi de cette géométrie. Les identifiants de groupe sont arbitraires.
