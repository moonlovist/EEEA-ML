## A2 · Comprendre l’idée

Le modèle affine prédit y_hat = w*x + b. La pente w s’exprime en min/km et l’ordonnée b en minutes. Le résidu est prédiction moins observation : e_i = y_hat_i - y_i. L’erreur absolue vaut |e_i|. MAE = (|e_1| + ... + |e_n|)/n, en minutes.

LinearRegression ajuste w et b en minimisant la somme des résidus d’entraînement au carré, avec une constante. L’évaluation emploie ici la MAE de validation. Le modèle de référence prédit la moyenne d’entraînement pour chaque trajet de validation. Ajustement et évaluation ont donc des rôles distincts. Modifier une droite manuelle change ses prédictions ; l’apprentissage estime les paramètres à partir des observations d’entraînement.

## A3 · Exemple guidé

```python
observed = np.array([12., 19., 32.])
predicted = np.array([10., 20., 30.])
absolute_errors = np.abs(predicted - observed)
print(absolute_errors, absolute_errors.mean())
from sklearn.linear_model import LinearRegression
sets = split_data("walks.csv")
X_train, y_train = sets["train"]
model = LinearRegression().fit(X_train, y_train)
print("w, b:", model.coef_[0], model.intercept_)
print("2 km:", model.predict([[2.]]))
```
